##### Data

```python
rho = 0.997
empty = {
    #Container,Expected_Volume(ml),Weight(g)
    "MC"   :[[0,mean([33.4249,33.4239,33.4240])]], #Measuring Cylinder
    "VF"   :[[0,mean([66.2980,66.2970,66.2980])]], #Volumetric Flask
    "Bk"   :[[0,mean([28.4140,28.4130,28.4130])]], #Beaker
    "NP"   :[[0,mean([28.4140,28.4130,28.4130])]], #Same as Beaker
    "GP"   :[[0,mean([28.4140,28.4130,28.4130])]], #Same as Beaker
    "Bk+TT":[[0,mean([40.5660,40.5650,40.5650])]], #Test-tube in beaker
}

full = {
    "GP"   :[
        [10,mean([38.3850,38.3860,38.3870])],
        [10,38.3317],
        [10,38.4670],
        ],
    "NP"   :[
        [10,mean([38.3780,38.3770,38.3760])],
        [10,mean([38.283,38.384,38.384])],
        [10,38.377],
        ],
    "VF"   :[
        [100,mean([165.543,165.541,165.539])],
        [100,mean([165.695,165.693,165.699])],
        [100,mean([165.477,165.476,165.477])],
        ],
    "MC"   :[
        [10,mean([43.523,43.525,43.523])],
        [10,43.4600],
        [10,43.4020],
    ],
    "Bk+TT":[
        [4,44.702],
        [4,44.783],
        [4,44.614],
    ],
}
```

##### Volume (mL) vs weights (g)

```python
water = {t : [x[:-1] + [round(x[-1]-empty[t][0][1],4)] for x in full[t]] for t in full}
for t in water:
    print(t,water[t])
```

```python
volume0 = {}
volume1 = {}
volume2 = {}
volume = {}
for t in water:
    volume0[t] = water[t][0][1]/rho
    volume1[t] = water[t][1][1]/rho
    volume2[t] = water[t][2][1]/rho
    volume[t] = water[t][0][0]
volume0 = Series(volume0)
volume1 = Series(volume1)
volume2 = Series(volume2)
volume = Series(volume)
```

##### Calculated Weights vs measured weights (g)

```python
for t in water:
    for x in water[t]:
        x[0] = round(x[0]*rho,4)
for t in water:
    print(t,water[t])
calc_weights = {t:water[t][0][0] for t in water}
```

##### Errors (in g)

```python
for t in water:
    for i in range(len(water[t])):
        water[t][i] = round(water[t][i][1] - water[t][i][0],3)
X = array(list(water.values()))
print(X)
df = DataFrame(X,index = water.keys(),columns = [0,1,2])
for t in water:
    print(t,water[t])
df
```

##### RMSE (g)

```python
rmse = {}
for t in water:
    rmse[t] = round(mean(array(water[t])**2)**0.5,3)
for t in water:
    print(t,round(rmse[t],3))
print("Total :",round(mean(array(list(rmse.values()))**2)**0.5,3))
df["RMSE"] = Series(rmse)
df["Mean Deviation"] = (df[0] + df[1] + df[2])/3
df["Mean Absolute deviation"] = (abs(df[0]) + abs(df[1]) + abs(df[2]))/3
df["V0"] = volume0
df["V1"] = volume1
df["V2"] = volume2
df["Mean Volume calculated from weight"] = (volume0 + volume1 + volume2)/3
df["desired volume"] = volume
df["calculated weight"] = Series(calc_weight)
df = round(df,3)
df
```
